```python
import pandas as pd

df = pd.read_csv("/kaggle/input/datasets/ahmedmohamed2003/cafe-sales-dirty-data-for-cleaning-training/dirty_cafe_sales.csv")
df.head()
```

```python
df.isna().sum()     # jumlah nilai kosong per kolom
```

```python
df['Quantity'].value_counts()      # ganti "Item" dengan nama kolom yang ada di datamu
```

```python
# ubah kolom yang seharusnya angka; nilai yang tidak bisa diubah jadi kosong
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

# ganti teks bermasalah dengan nilai kosong
df = df.replace(["ERROR", "UNKNOWN"], pd.NA)

# hapus baris yang masih kosong di kolom penting
df = df.dropna(subset=["Quantity"])
```

```python
df.groupby("Item")["Quantity"].sum().sort_values(ascending=False)
```
