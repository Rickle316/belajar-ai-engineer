```python
import pandas as pd

df = pd.read_csv('/kaggle/input/datasets/lainguyn123/student-performance-factors/StudentPerformanceFactors.csv')
df.head()
```

```python
df.shape            # jumlah baris dan kolom
df.info()           # nama kolom, tipe data, jumlah nilai terisi
df.describe()       # ringkasan statistik kolom angka
df.isna().sum()     # jumlah nilai kosong per kolom
```

```python
print(df.shape)
df.describe()
```

```python
(df.isna().mean() * 100).round(2)
```

```python
df_bersih = df.copy()
```

```python
for kol in ["Teacher_Quality", "Parental_Education_Level", "Distance_from_Home"]: 
    df_bersih[kol] = df_bersih[kol].fillna(df_bersih[kol].mode()[0])
```

```python
df_bersih.isna().sum()
```

```python
df_bersih.duplicated().sum()
```

```python
df_bersih.describe()
df_bersih[df_bersih['Exam_Score'] > 100]
(df_bersih['Attendance']>100).sum()
```

```python
df_bersih['Parental_Involvement'].value_counts()
```

```python
df_bersih.groupby("Parental_Involvement")["Exam_Score"].mean()
```

mencari nilai rata2 korelasi nilai dengan parental involvement

```python
urutan = ["Low", "Medium", "High"]
df_bersih.groupby("Parental_Involvement")["Exam_Score"].mean().reindex(urutan)
```

cara untuk mensortir sesuai dengan urutan

```python
df_bersih.groupby("Tutoring_Sessions")["Exam_Score"].agg(["mean","count"])
```

```python
kolom_angka = ["Hours_Studied", "Attendance", "Previous_Scores", "Sleep_Hours", "Tutoring_Sessions","Physical_Activity", "Exam_Score"]
df_bersih[kolom_angka].corr()["Exam_Score"].sort_values(ascending=False)
```

Korelasi antara hasil ujian dengan kehadiran dan jam belajar

```python
df_bersih["lulus"] = df_bersih["Exam_Score"] >+ 70
df_bersih.groupby('School_Type')['lulus'].mean() * 100
```

berapa persentase rata2 siswa sekolah swasta dan negeri yang lulus dengan nilai >= 70

```python
urutan = ["Low", "Medium", "High"]
df_bersih.pivot_table(values="Exam_Score", index="Parental_Involvement", columns="Teacher_Quality", aggfunc="mean").reindex(urutan)
```
